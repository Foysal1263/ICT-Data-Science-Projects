```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn import metrics

# 1. Setup Visuals
sns.set_style('whitegrid')

# 2. Define Features (X) and Target (y)
# We exclude non-numeric columns like Email and Address
X = customers[['Avg. Session Length', 'Time on App', 'Time on Website', 'Length of Membership']]
y = customers['Yearly Amount Spent']

# 3. Split the data
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=101)

# 4. Train the Model
lm = LinearRegression()
lm.fit(X_train, y_train)

# 5. Output the Mathematical Results
print(f"Intercept: {lm.intercept_:.2f}\n")
coeff_df = pd.DataFrame(lm.coef_, X.columns, columns=['Coefficient'])
print(coeff_df)

# 6. Perfect Visualization (The "Proof" Plot)
predictions = lm.predict(X_test)
plt.figure(figsize=(10,6))
sns.scatterplot(x=y_test, y=predictions)
plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], color='red', lw=2) # Perfect fit line
plt.xlabel('Y Test (True Values)')
plt.ylabel('Predicted Y')
plt.title('Model Accuracy: Predicted vs. Actual')
plt.show()
```

```python
import pandas as pd
from matplotlib import pyplot as plt
%matplotlib inline
```

```python
df= pd.read_csv('insurance_data.csv')
```

```python
df.head()
```

```python
plt.scatter(df['age'], df['bought_insurance'],color='red')
plt.xlabel('Age')
plt.ylabel('Bought Insurance')
plt.title('Age vs Bought Insurance')
plt.show()
```

```python
plt.scatter(df['age'], df['bought_insurance'],marker='*', color='red')
plt.xlabel('Age')
plt.ylabel('Bought Insurance')
plt.title('Age vs Bought Insurance')
plt.show()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
```

```python
X_train, X_test, y_train, y_test = train_test_split(df[['age']], df.bought_insurance, test_size=0.2)
```

```python
X_test
```

```python
y_test
```

```python
logestic_model = LogisticRegression()
logestic_model.fit(X_train, y_train)
print(logestic_model.coef_)
```

```python
y_predict = logestic_model.predict(X_test)
```

```python
y_predict
```

```python
logestic_model.predict_proba(X_test)
```

```python
logestic_model.score(X_test, y_test)
```

```python
logestic_model.coef_
```

```python
logestic_model.intercept_
```

```python
import math
def sigmoid(x):
    return 1 / (1 + math.exp(-x))
```

```python
def prediction_function(age):
    linear_model = logestic_model.intercept_ + logestic_model.coef_ * age
    return sigmoid(linear_model)
```

```python
age = 29
prediction_function(age)
```

```python
prediction_function(58)
```
