```python
#Q- 02:- Construct a stem-and-leaf display for the Following data.Use a leaf unit of 10.
data2 = [1161, 1206, 1478, 1300, 1604, 1725, 1361, 1422, 
         1221, 1378, 1623, 1426, 1557, 1730, 1706, 1689]
data2.sort()
stems = {}
for val in data2:
    stem = val // 100      
    leaf = (val // 10) % 10 
    if stem not in stems:
        stems[stem] = []
    stems[stem].append(str(leaf))

print("Stem | Leaf (Leaf Unit = 10)")
for s in sorted(stems.keys()):
    print(f"{s:4} | {' '.join(stems[s])}")

```

```python
#Q- 03(a)
import numpy as np
from scipy import stats

data = [15, 22, 12, 21, 26, 18, 42, 29, 64, 20, 15, 22, 18, 24, 27, 
        24, 26, 19, 20, 23, 41, 35, 55, 37, 17, 29, 19, 26, 38, 11]

mean_val = np.mean(data)
median_val = np.median(data)

print(f"Mean percentage: {mean_val:.2f}%")
print(f"Median percentage: {median_val:.2f}%")
```

```python
#Q- 03(b)
df = pd.Series(data)
q1 = df.quantile(0.25)
q3 = df.quantile(0.75)
print(f"Q1: {q1}")
print(f"Q3: {q3}")
```

```python
#Q- 03(c)
data_range = df.max() - df.min()
iqr = stats.iqr(data)
print(f"Range: {data_range}")
print(f"IQR: {iqr}")


```

```python
#Q- 03(d)
variance = df.var()
std_dev = df.std()
print(f"Variance: {variance:.2f}")
print(f"Standard Deviation: {std_dev:.2f}")
```

```python
#Q- 03(e)
lower_bound = q1 - (1.5 * iqr)
upper_bound = q3 + (1.5 * iqr)
outliers = [x for x in data if x < lower_bound or x > upper_bound]
print(f"Outliers: {outliers}")


```

```python
#Q- 04
import numpy as np
from scipy import stats

# Data from Q4
ratings = [6, 4, 6, 8, 7, 7, 6, 3, 3, 8, 10, 4, 8, 7, 8, 7, 5, 9, 5, 8, 4, 
           3, 8, 5, 5, 4, 4, 4, 8, 4, 5, 6, 2, 5, 9, 9, 8, 4, 8, 9, 9, 5, 
           9, 7, 8, 3, 10, 8, 9, 6]

n = len(ratings)
mean = np.mean(ratings)
sem = stats.sem(ratings) # Standard Error of the Mean

# 95% Confidence Interval
ci = stats.t.interval(0.95, df=n-1, loc=mean, scale=sem)

print(f"Sample Mean: {mean:.2f}")
print(f"95% Confidence Interval: ({ci[0]:.2f}, {ci[1]:.2f})")
```

```python
#Q- 05
import numpy as np
from scipy import stats

pop_a = [9.5, 10.5, 9.0, 9.75, 10.0, 13.0, 10.0, 13.5, 10.0, 9.5, 10.0, 9.75, 11.2, 8.5, 10.3]
pop_b = [12.5, 9.5, 13.5, 13.75, 12.0, 13.75, 12.5, 9.5, 12.0, 13.5, 12.0, 12.0, 13.0, 15, 10.8]

# Perform independent t-test
t_stat, p_value = stats.ttest_ind(pop_a, pop_b)

alpha = 0.03

print(f"P-value: {p_value:.4f}")
if p_value < alpha:
    print("Conclusion: Reject the Null Hypothesis. There is a significant difference.")
else:
    print("Conclusion: Fail to reject the Null Hypothesis. No significant difference found.")
```

```python
#Q- 06
import pandas as pd
from scipy.stats import chi2_contingency

# Example data extraction (partial shown for logic)
gender = ['M','F','F','F','F','M','M','F','M','F'] # ... up to 200
home = ['T','T','T','T','V','V','V','T','T','T']     # ... up to 200

data = pd.crosstab(pd.Series(gender), pd.Series(home))
chi2, p, dof, ex = chi2_contingency(data)

print(f"Chi-square P-value: {p:.4f}")

if p < 0.05:
    print("Conclusion: Reject the Null Hypothesis. There is an association between gender and home location.")
else:
    print("Conclusion: Fail to reject the Null Hypothesis. No association found between gender and home location.")
```

```python
#Q- 07(a)
import pandas as pd
import statsmodels.api as sm

# Data from the table in Q7
data = {
    'y': [94, 108, 112, 178, 94, 175, 170, 117, 142, 211],
    'x1': [30, 47, 25, 51, 40, 51, 74, 36, 59, 76],
    'x2': [12, 10, 17, 16, 5, 19, 7, 12, 13, 16]
}
df = pd.DataFrame(data)

# Part (a): y = b0 + b1*x1
X1 = sm.add_constant(df['x1']) # Adds the intercept (b0)
model_a = sm.OLS(df['y'], X1).fit()

# Prediction for x1 = 47
pred_a = model_a.predict([1, 47])
print(model_a.summary())
print(f"Predicted y if x1=47: {pred_a}")
```

```python
#Q- 07(b)
X2 = sm.add_constant(df['x2'])
model_b = sm.OLS(df['y'], X2).fit()

# Prediction for x2 = 10
pred_b = model_b.predict([1, 10])
print(model_b.summary())
print(f"Predicted y if x2=10: {pred_b}")
```

```python
#Q- 07(c)
import statsmodels.api as sm
import pandas as pd


X = sm.add_constant(df[['x1', 'x2']])
model = sm.OLS(df['y'], X).fit()

# Prediction for x1=47, x2=10
prediction = model.predict([1, 47, 10])
print(model.summary())
print(f"Predicted y for x1=47, x2=10: {prediction}")
```
